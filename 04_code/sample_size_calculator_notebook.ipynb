# 📊 Population Sample Size Calculator
This notebook calculates required sample sizes based on statistical confidence levels, margin of error, and population size.

## 🔢 Formula Overview

- Infinite population sample size:

$$ m = \frac{Z^2 \cdot p \cdot (1 - p)}{E^2} $$

- Finite population correction:

$$ n = \frac{m}{1 + \frac{m - 1}{N}} $$

```python

import math

def calculate_sample_size(z, p, e, N=None):
    # Calculate m (infinite population)
    m = (z**2 * p * (1 - p)) / (e**2)

    # If N (finite population) is provided, adjust sample size
    if N:
        n = m / (1 + ((m - 1) / N))
    else:
        n = m

    return math.ceil(n)

# Example usage
z = 1.96  # 95% confidence
p = 0.5   # estimated proportion
e = 0.05  # margin of error
N = 1000  # population size

sample_size = calculate_sample_size(z, p, e, N)
print(f"Required sample size: {sample_size}")
```
