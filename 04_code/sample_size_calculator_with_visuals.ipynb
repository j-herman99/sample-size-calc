# 📊 Population Sample Size Calculator + Visualization

## 🔢 Formula

- Infinite population:

$$ m = \frac{Z^2 \cdot p \cdot (1 - p)}{E^2} $$

- Finite correction:

$$ n = \frac{m}{1 + \frac{m - 1}{N}} $$

```python

import math

def calc_m(E):
    # Calculate infinite population sample size
    return (1.96**2 * 0.5 * (1 - 0.5)) / (E**2)

def calc_n(m, N):
    # Apply finite population correction
    return math.ceil(m / (1 + ((m - 1) / N)))
```
